```python
import sys
from pathlib import Path

ml_root = Path.cwd().resolve()
if ml_root.name != "ml" and (ml_root / "ml").is_dir():
    ml_root = ml_root / "ml"
if str(ml_root) not in sys.path:
    sys.path.insert(0, str(ml_root))

from src.config import load_config
from src.data.loader import load_jsonl
from src.evaluation import PersonaEvaluator

cfg = load_config()
evaluator = PersonaEvaluator()
print("PersonaEvaluator ready.")
```

## 1. Evaluate Sample Output vs Reference Persona Responses

```python
sample_pairs = [
    {
        "prompt": "Arre bhai, weekend ka kya scene hai?",
        "generated": "Kuch nahi yaar, bas chill karenge ghar pe 😂",
        "reference": "Kuch nahi yaar, bas socha tha chill karenge... par kaam ne aisi taisi kar di 😭"
    },
    {
        "prompt": "Chai peene chalte hain niche?",
        "generated": "Haan chal tapri pe chalte hain ☕",
        "reference": "Chai ke liye mana kab kiya hai maine? Chal 2 min me aaya tapri pe ☕"
    }
]

report = evaluator.evaluate(sample_pairs)
print(report.summary_text())
```

## 2. Inspect Detailed Per-Sample Scores

```python
for idx, sample in enumerate(report.samples, 1):
    print(f"Sample {idx}:")
    print(f"  Prompt   : {sample['prompt']}")
    print(f"  Generated: {sample['generated']}")
    print(f"  Reference: {sample['reference']}")
    print(f"  Metrics  : {sample['metrics']}")
    print()
```
