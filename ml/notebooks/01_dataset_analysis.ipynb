**Output:** Dataset statistics summary, stylistic signatures, analysis report JSON

```python
import sys
from pathlib import Path

# Add ml root to path
ml_root = Path.cwd().resolve()
if ml_root.name != "ml" and (ml_root / "ml").is_dir():
    ml_root = ml_root / "ml"
if str(ml_root) not in sys.path:
    sys.path.insert(0, str(ml_root))

from src.config import load_config
from src.analysis import DatasetAnalyzer, analyze_dataset

cfg = load_config()
print("Loaded configuration for environment:", cfg.environment)
```

## 1. Run Reusable Dataset Analyzer

```python
dataset_path = Path(cfg.paths.sample_data) / "sample_conversations.jsonl"
print("Analyzing dataset:", dataset_path)

analyzer = DatasetAnalyzer(target_speaker="persona")
report = analyzer.analyze_file(dataset_path)

print(report.summary_text())
```

## 2. Inspect Speaker Turn Dynamics

```python
print("Total Conversations:", report.total_conversations)
print("Total Messages     :", report.total_messages)
print("Messages by Speaker:", report.messages_by_speaker)
print("Avg Words / Message:", report.avg_message_length_words)
print("Speaker Avg Chars  :", report.speaker_avg_lengths)
```

## 3. Inspect Stylistic Markers (Emojis, Hinglish, Punctuation)

```python
print(f"Question Frequency   : {report.question_rate:.1%}")
print(f"Exclamation Frequency: {report.exclamation_rate:.1%}")
print(f"Emoji Presence Rate  : {report.emoji_message_ratio:.1%}")
print("Top Emojis Used      :", report.top_emojis)
print(f"Hinglish Rate        : {report.hinglish_ratio:.1%}")
print("Top Bigrams (Phrases):", report.top_bigrams[:5])
```
