# Persona Engine ML Lab

**Environment:** Google Colab / Local
**Purpose:** Run baseline zero-shot / few-shot inference to benchmark foundation model capability prior to fine-tuning.
**Input:** Prompt dialogues, base foundation model
**Output:** Baseline generated responses, comparison with reference persona style

```python
import sys
from pathlib import Path

ml_root = Path.cwd().resolve()
if ml_root.name != "ml" and (ml_root / "ml").is_dir():
    ml_root = ml_root / "ml"
if str(ml_root) not in sys.path:
    sys.path.insert(0, str(ml_root))

from src.config import load_config
from src.inference import PersonaGenerator, GenerationParams

cfg = load_config()
print("Base Model Target:", cfg.model.name)
```

## 1. Initialize Inference Generator

```python
generator = PersonaGenerator(base_model_name_or_path=cfg.model.name)
generator.load_model()
print("Generator ready. Model loaded state:", generator.is_model_loaded)
```

## 2. Test Conversation Prompt Generation

```python
test_messages = [
    {"speaker": "user", "text": "Arre bhai, weekend ka kya scene hai?"}
]

params = GenerationParams(temperature=0.7, max_new_tokens=128)
response = generator.generate(test_messages, params=params)
print("User Prompt      :", test_messages[0]["text"])
print("Generated Output :", response)
```
