```python
import sys
from pathlib import Path

ml_root = Path.cwd().resolve()
if ml_root.name != "ml" and (ml_root / "ml").is_dir():
    ml_root = ml_root / "ml"
if str(ml_root) not in sys.path:
    sys.path.insert(0, str(ml_root))

from src.annotation import MessageAnnotation, AnnotationValidator, create_annotated_conversation
from src.data.loader import load_jsonl

print("Annotation models and validator imported successfully.")
```

## 1. Explore MessageAnnotation Schema

```python
ann = MessageAnnotation(
    topic="weekend_plans",
    intent="joke",
    emotion="happy",
    tone="playful",
    humor=True,
    sarcasm=False,
    teasing=True,
    language="hinglish",
    language_mixing=True,
    response_strategy="TEASE",
    relationship_context="close_friend"
)
print("Sample Annotation Dict:")
print(ann.to_dict())
```

## 2. Validate Annotated Conversation Structure

```python
annotated_sample = create_annotated_conversation(
    conversation_id="annotated-demo-001",
    persona_id="default",
    messages=[
        {
            "speaker": "user",
            "text": "Weekend ka kya plan hai?",
            "timestamp": "2026-10-01T12:00:00Z"
        },
        {
            "speaker": "persona",
            "text": "Kuch nahi bhai, bas sona hai 😂",
            "timestamp": "2026-10-01T12:01:00Z",
            "annotations": ann.to_dict()
        }
    ]
)

validator = AnnotationValidator()
validator.validate(annotated_sample)
print("Annotated conversation successfully validated against annotation.schema.json!")
```
